```python
import pandas as pd
import json

# 1. Datei korrekt mit Komma als Trennzeichen einlesen
try:
    df_raw = pd.read_csv('/content/short_term_traffic_1911_1230_1400.csv', sep=',')
    print("Datei erfolgreich eingelesen!")
except Exception as e:
    print(f"Fehler: {e}")

# 2. JSON-Payload parsen und in eigene Spalten aufteilen
def parse_payload(val):
    try:
        return json.loads(val)
    except:
        return {}

payload_parsed = df_raw['payload'].apply(parse_payload)
df_payload = pd.json_normalize(payload_parsed)

# 3. Zusammenführen von 'permanent_id' und den neuen Spalten
df_final = pd.concat([df_raw[['permanent_id']], df_payload], axis=1)

# 4. Vorschau anzeigen
print("Vorschau der aufbereiteten Daten:")
display(df_final.head())

# 5. Als Excel-Datei (.xlsx) speichern
output_path = '/content/short_term_traffic_aufbereitet.xlsx'
df_final.to_excel(output_path, index=False)
print(f"\nDatei erfolgreich gespeichert unter: {output_path}")
print("Sie können diese Datei nun über die Dateileiste links herunterladen.")
```
